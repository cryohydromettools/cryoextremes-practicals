```python
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import pandas as pd

# Configurar estilo visual padrão
plt.style.use(
    'seaborn-v0_8-whitegrid'
    if 'seaborn-v0_8-whitegrid' in plt.style.available
    else 'default'
)

# Definir dicionário de cores fixas por fonte de dados
colors = {
    'wrf': 'green',  # verde
    'tower': '#d62728',  # Vermelho
    'buoy': '#1f77b4',  # Laranja
}

# Criar a figura
fig, axes = plt.subplots(
    nrows=6, ncols=1, figsize=(14, 10), sharex=True, dpi=100
)

# 1. Temperatura do Ar
axes[0].plot(
    df_wrf.index, df_wrf['T2'], label='WRF', color=colors['wrf'], lw=1.2
)

axes[0].plot(
    df_tower.index,
    df_tower['tair'],
    label='Tower',
    color=colors['tower'],
    lw=1.2,
)
axes[0].plot(
    df_bouy.index, df_bouy['tair'], label='Buoy', color=colors['buoy'], lw=1.2
)
axes[0].set_ylabel('Air Temp (°C)', fontsize=10, fontweight='bold')
axes[0].grid(True, linestyle='--', alpha=0.5)

# 2. Pressão Atmosférica
axes[1].plot(
    df_wrf.index, df_wrf['SLP'], label='WRF', color=colors['wrf'], lw=1.2
)
axes[1].plot(
    df_tower.index,
    df_tower['press'],
    label='Tower',
    color=colors['tower'],
    lw=1.2,
)
axes[1].plot(
    df_bouy.index, df_bouy['press'], label='Buoy', color=colors['buoy'], lw=1.2
)
axes[1].set_ylabel('Pressure (hPa)', fontsize=10, fontweight='bold')
axes[1].grid(True, linestyle='--', alpha=0.5)

# 3. Umidade Relativa
axes[2].plot(
    df_wrf.index, df_wrf['RH2'], label='WRF', color=colors['wrf'], lw=1.2
)

axes[2].plot(
    df_tower.index, df_tower['rh'], label='Tower', color=colors['tower'], lw=1.2
)
axes[2].plot(
    df_bouy.index, df_bouy['rh'], label='Buoy', color=colors['buoy'], lw=1.2
)
axes[2].set_ylabel('RH (%)', fontsize=10, fontweight='bold')
axes[2].grid(True, linestyle='--', alpha=0.5)

# 4. Radiação Solar Incident (SWin)
axes[3].plot(
    df_wrf.index, df_wrf['SWDOWN'], label='WRF', color=colors['wrf'], lw=1.2
)
axes[3].plot(
    df_tower.index,
    df_tower['SUp'],
    label='Tower',
    color=colors['tower'],
    lw=1.2,
)
axes[3].plot(
    df_bouy.index, df_bouy['SUp'], label='Buoy', color=colors['buoy'], lw=1.2
)
axes[3].set_ylabel('SW In (W/m²)', fontsize=10, fontweight='bold')
axes[3].grid(True, linestyle='--', alpha=0.5)

# 5. Wind speed (SWin)

axes[4].plot(
    df_wrf.index, df_wrf['WS10'], label='WRF', color=colors['wrf'], lw=1.2
)

axes[4].plot(
    df_bouy.index, df_bouy['ws'], label='Buoy', color=colors['buoy'], lw=1.2
)
axes[4].set_ylabel('WS (m/s)', fontsize=10, fontweight='bold')
axes[4].grid(True, linestyle='--', alpha=0.5)

# 7. Precipitação
axes[5].plot(
    df_wrf.index,
    df_wrf['PRECIP_RATE'],
    label='WRF',
    color=colors['wrf'],
    lw=1.2,
)

axes[5].set_ylabel('Precip (mm/h)', fontsize=10, fontweight='bold')
axes[5].set_xlabel('Date', fontsize=11, fontweight='bold')
axes[5].grid(True, linestyle='--', alpha=0.5)

# DEFINIR O LIMITE DO EIXO X (xlim)
axes[5].set_xlim(pd.to_datetime('2024-12-06'), pd.to_datetime('2024-12-13'))

# Formatação das datas no Eixo X
axes[5].xaxis.set_major_formatter(mdates.DateFormatter('%b %d\n%Y'))
axes[5].xaxis.set_major_locator(mdates.AutoDateLocator())

# Legenda única no topo para toda a figura
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    loc='upper center',
    bbox_to_anchor=(0.5, 0.98),
    ncol=4,
    frameon=True,
    fontsize=11,
)

plt.tight_layout()
plt.subplots_adjust(top=0.93)  # Reserva espaço no topo para a legenda principal

fig.savefig('fig/comp_torre_boia.png', dpi=300, facecolor='w', bbox_inches='tight', pad_inches=0.1)

plt.show()

```

```python
file_path = '/home/lacrio/cryoextremes-practicals/data/EP/Pinguim_EP/eddypro_PG_OP43_full_output_2025-04-22T110846_exp.csv'

# Read file
df = pd.read_csv(
    file_path,
    sep=",",
    skiprows=[0, 2],
    na_values=-9999
)
print(df.columns)
```

```python
# Select variables of interest
df_sel = df[['date', 'time', 'H', 'LE', 'air_temperature', 'RH', 'wind_speed']]

# Create datetime index
df_sel["datetime"] = pd.to_datetime(df_sel["date"] + " " + df_sel["time"])
df_sel['datetime'] = df_sel['datetime'] + pd.Timedelta(hours=3)
df_sel = df_sel.set_index("datetime")
df_sel = df_sel.drop(columns=["date", "time"])
df_sel = df_sel.loc['2024-12-06': '2024-12-13'].resample('1h').mean()
df_sel
```

```python
cols_tower = ['SUp', 'SDn', 'LUpCo', 'LDnCo', 'SHF']
cols_eddy = ['H', 'LE']

df_merged = pd.merge(
    df_tower[cols_tower],
    df_sel[cols_eddy],
    on='datetime',
    how='inner'  # Usa 'outer' si quieres conservar todas las horas de ambos
)
df_merged
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Eliminar filas con NaN
df_clean = df_merged.dropna().copy()

# 2. CÁLCULOS DE LOS COMPONENTES DE RADIACIÓN Y SEB
df_clean['Net_Shortwave'] = df_clean['SUp'] + df_clean['SDn']
df_clean['Net_Longwave'] = df_clean['LUpCo'] + df_clean['LDnCo']
df_clean['Net_Radiation_Rn'] = (
    df_clean['Net_Shortwave'] + df_clean['Net_Longwave']
)

df_clean['Soil_Heat_Flux_G'] = df_clean['SHF']
df_clean['Sensible_Heat_Flux_H'] = df_clean['H']
df_clean['Latent_Heat_Flux_LE'] = df_clean['LE']

df_clean['SEB_Residual'] = (
    df_clean['Net_Radiation_Rn']
    - df_clean['Sensible_Heat_Flux_H']
    - df_clean['Latent_Heat_Flux_LE']
    - df_clean['Soil_Heat_Flux_G']
)

# 3. GRAFICAR EN 3 SUBPLOTS
fig, (ax1, ax2, ax3) = plt.subplots(
    3, 1, figsize=(15, 12), sharex=True, dpi=120
)

# --- PANEL 1: Componentes de Radiación ---
ax1.plot(
    df_clean.index,
    df_clean['SUp'],
    label='Shortwave Upward ($SW_{up}$)',
    color='#ff7f0e',
    linestyle='--',
    linewidth=1.2,
)
ax1.plot(
    df_clean.index,
    df_clean['SDn'],
    label='Shortwave Downward ($SW_{dn}$)',
    color='#d62728',
    linestyle='--',
    linewidth=1.2,
)
ax1.plot(
    df_clean.index,
    df_clean['LUpCo'],
    label='Longwave Upward ($LW_{up}$)',
    color='#9467bd',
    linestyle=':',
    linewidth=1.2,
)
ax1.plot(
    df_clean.index,
    df_clean['LDnCo'],
    label='Longwave Downward ($LW_{dn}$)',
    color='#8c564b',
    linestyle=':',
    linewidth=1.2,
)
ax1.plot(
    df_clean.index,
    df_clean['Net_Radiation_Rn'],
    label='Net Radiation ($R_n$)',
    color='black',
    linewidth=2.0,
)

ax1.set_title(
    '1. Radiation Components & Net Radiation ($R_n$)',
    fontsize=12,
    fontweight='bold',
)
ax1.set_ylabel('Radiation ($W \\cdot m^{-2}$)', fontsize=11)
ax1.axhline(0, color='gray', linestyle=':', alpha=0.6)
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.legend(loc='upper right', frameon=True, facecolor='white', fontsize=9, ncol=2)

# --- PANEL 2: Flujos Turbulentos (H, LE) y Calor en el Suelo (G) ---
ax2.plot(
    df_clean.index,
    df_clean['Sensible_Heat_Flux_H'],
    label='Sensible Heat Flux ($H$)',
    color='#d95f02',
    linewidth=1.8,
)
ax2.plot(
    df_clean.index,
    df_clean['Latent_Heat_Flux_LE'],
    label='Latent Heat Flux ($LE$)',
    color='#7570b3',
    linewidth=1.8,
)
ax2.plot(
    df_clean.index,
    df_clean['Soil_Heat_Flux_G'],
    label='Soil Heat Flux ($G$ / $SHF$)',
    color='#2ca02c',
    linewidth=1.8,
)

ax2.set_title(
    '2. Turbulent Heat Fluxes ($H$, $LE$) & Soil Heat Flux ($G$)',
    fontsize=12,
    fontweight='bold',
)
ax2.set_ylabel('Heat Flux ($W \\cdot m^{-2}$)', fontsize=11)
ax2.axhline(0, color='gray', linestyle=':', alpha=0.6)
ax2.grid(True, linestyle='--', alpha=0.5)
ax2.legend(loc='upper right', frameon=True, facecolor='white', fontsize=9.5)
ax2.set_ylim(-150, 150)

# --- PANEL 3: Residuo del Balance de Energía ---
ax3.plot(
    df_clean.index,
    df_clean['SEB_Residual'],
    label='Energy Balance Residual ($R_n - H - LE - G$)',
    color='#e7298a',
    linestyle='--',
    linewidth=1.8,
)

ax3.set_title(
    '3. Surface Energy Balance Residual (Imbalance)',
    fontsize=12,
    fontweight='bold',
)
ax3.set_xlabel('Date and Time (UTC)', fontsize=11, labelpad=8)
ax3.set_ylabel('Residual ($W \\cdot m^{-2}$)', fontsize=11)
ax3.axhline(0, color='black', linestyle=':', alpha=0.8)
ax3.grid(True, linestyle='--', alpha=0.5)
ax3.legend(loc='upper right', frameon=True, facecolor='white', fontsize=9.5)

plt.gcf().autofmt_xdate()
plt.tight_layout()
plt.show()
```
