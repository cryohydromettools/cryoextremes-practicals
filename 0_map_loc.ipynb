## 1. Import Libraries & Setup Map Projection

We load Cartopy and Matplotlib to configure the geospatial canvas.

```python
import cartopy.crs as ccrs
import matplotlib.pyplot as plt
```

--- 
## 2. Generate Global Map and Plot Study Sites

We create the figure using a Robinson projection and overlay the specific coordinates for each study location with custom markers and labels.

```python
proj_data = ccrs.PlateCarree()

fig = plt.figure(figsize=[12, 6])

ax1 = plt.subplot(1, 1, 1, projection=ccrs.Robinson())
ax1.coastlines('50m', zorder=2)
ax1.stock_img()

ax1.plot(10.77, 46.80, '*r', markersize= 20, transform=proj_data, label='Hintereisferner Glacier')

ax1.plot(-68.14, -16.27, '*k', markersize= 20, transform=proj_data, label='Zongo Glacier')

ax1.plot(-52.09, -32.13, '*b', markersize= 20, transform=proj_data, label='Rio Grande')

ax1.plot(-57.93, -62.09, '*g', markersize= 20, transform=proj_data,  label='Antarctic Peninsula')

plt.legend(loc='lower right')
fig.savefig('fig/Fig01.png', dpi=300, facecolor='w', bbox_inches='tight', pad_inches=0.1)
```
